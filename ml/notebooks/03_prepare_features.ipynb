```python
import numpy as np
import pandas as pd

hog = np.load("../models/hog_pca.npy")
lbp = np.load("../models/lbp_pca.npy")
labels = np.load("../models/labels.npy")

df_hog = pd.DataFrame(hog)
df_hog["label"] = labels

df_lbp = pd.DataFrame(lbp)
df_lbp["label"] = labels

df_hog.to_csv("../models/hog_pca.csv", index=False)
df_lbp.to_csv("../models/lbp_pca.csv", index=False)
```
