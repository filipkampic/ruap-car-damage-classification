```python
import os
import matplotlib.pyplot as plt
from PIL import Image
import numpy as np

DATASET_PATH = "../dataset"

classes = ["minor", "moderate", "severe"]

counts = {cls: len(os.listdir(os.path.join(DATASET_PATH, cls))) for cls in classes}
counts
```

```python
plt.bar(counts.keys(), counts.values(), color=["green", "orange", "red"])
plt.title("Distribucija klasa u datasetu")
plt.xlabel("Klasa")
plt.ylabel("Broj slika")
plt.show()
```

```python
fig, axs = plt.subplots(3, 3, figsize=(10, 10))

for i, cls in enumerate(classes):
    folder = os.path.join(DATASET_PATH, cls)
    images = os.listdir(folder)[:3]
    for j, img_name in enumerate(images):
        img = Image.open(os.path.join(folder, img_name))
        axs[i, j].imshow(img)
        axs[i, j].set_title(f"{cls}")
        axs[i, j].axis("off")
plt.show()
```

```python
sizes = []

for cls in classes:
    folder = os.path.join(DATASET_PATH, cls)
    for img_name in os.listdir(folder)[:50]:
        img = Image.open(os.path.join(folder, img_name))
        sizes.append(img.size)
sizes[:10]
```

```python
widths = [s[0] for s in sizes]
heights = [s[1] for s in sizes]

print("Prosječna širina: ", np.mean(widths))
print("Prosječna visina: ", np.mean(heights))
```

```python
plt.hist(widths, bins=20, alpha=0.7, label="Širina")
plt.hist(heights, bins=20, alpha=0.7, label="Visina")
plt.legend()
plt.title("Distribucija dimenzija slika")
plt.show()
```
